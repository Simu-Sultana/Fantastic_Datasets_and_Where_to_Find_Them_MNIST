```python
# Results Visualization for MNIST Handwritten Digit Recognition

import matplotlib.pyplot as plt
```

```python
# CNN training results
# Replace these values with the real values from your CNN training output if needed.

epochs = [1, 2, 3, 4, 5]

training_accuracy = [0.95, 0.98, 0.985, 0.990, 0.993]
validation_accuracy = [0.96, 0.975, 0.982, 0.987, 0.990]

training_loss = [0.18, 0.07, 0.045, 0.030, 0.022]
validation_loss = [0.13, 0.08, 0.055, 0.040, 0.035]
```

```python
# Plot CNN accuracy

plt.figure(figsize=(8, 5))

plt.plot(epochs, training_accuracy, marker='o', label='Training Accuracy')
plt.plot(epochs, validation_accuracy, marker='o', label='Validation Accuracy')

plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.title('CNN Training and Validation Accuracy')
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Plot CNN loss

plt.figure(figsize=(8, 5))

plt.plot(epochs, training_loss, marker='o', label='Training Loss')
plt.plot(epochs, validation_loss, marker='o', label='Validation Loss')

plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('CNN Training and Validation Loss')
plt.legend()
plt.grid(True)

plt.show()
```

```python
# Compare KNN and CNN accuracy
# Replace these with your real accuracy values.

models = ['KNN', 'CNN']
accuracy = [0.97, 0.99]

plt.figure(figsize=(6, 5))

plt.bar(models, accuracy)

plt.xlabel('Model')
plt.ylabel('Accuracy')
plt.title('MNIST Model Accuracy Comparison')
plt.ylim(0.90, 1.00)
plt.grid(axis='y')

plt.show()
```
