```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
import tensorflow as tf
import seaborn as sns
```

```python

mnist = tf.keras.datasets.mnist
(X_train_full, y_train_full), (X_test, y_test) = mnist.load_data()

print(y_train_full.dtype, "-", y_train_full.shape)
print(X_test.dtype, "-", X_test.shape)
print(y_test.dtype, "-", y_test.shape)

X_train_full[30]
y_train_full[30]
```

```python
X_valid, X_train = X_train_full[:5000], X_train_full[5000:]
y_valid, y_train = y_train_full[:5000], y_train_full[5000:]

train_indices = np.random.randint(0, X_train.shape[0], 2)
valid_indices = np.random.randint(0, X_valid.shape[0], 2)

plt.figure(figsize=(6, 6))

for i in range(2):
    plt.subplot(2, 2, i + 1)  # 2 rows, 2 columns
    plt.imshow(X_train[train_indices[i]], cmap="gray")
    plt.axis("off")
    plt.title(f"Train {train_indices[i]}")

for i in range(2):
    plt.subplot(2, 2, i + 3)  # Second row, 2 columns
    plt.imshow(X_valid[valid_indices[i]], cmap="gray")
    plt.axis("off")
    plt.title(f"Valid {valid_indices[i]}")

plt.tight_layout()
plt.show()
```
